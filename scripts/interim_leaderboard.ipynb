# Interim leaderboards

Builds, checks and publishes the interim leaderboards, after `INTERIM_LEADERBOARD_DEADLINE`. Run it from `scripts/`, in the backend environment, with AWS credentials for the challenge bucket. See `scripts/README.md`.

```python
import sys
from pathlib import Path

# Run from scripts/: put the repo root on the path for `backend` and `scripts`.
REPO_ROOT = Path.cwd().resolve().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from scripts import leaderboard_release as release  # noqa: E402

STAGE = "interim"
```

## 1. Build

Builds every track's leaderboards for this stage, from the submissions made before the deadline in `backend/config.py`. Building a stage's boards needs the bootstrap scores, so this can take a few minutes.

```python
built = release.build(STAGE)
```

## 2. Check

Checks each track's master leaderboard. **BLOCK** items stop the upload; **REVIEW** items are for you to look at. The table compares the top of the board with the live leaderboard.

```python
result = release.check(STAGE, built)
```

## 3. Save a local copy

CSV files in `scripts/output/` (gitignored: they include participants' details), for review and for blog posts.

```python
release.save_local(STAGE, built)
```

## 4. Upload

Writes each board to where the Space reads it (`leaderboard/{stage}/{track}/{slug}_leaderboard_latest.csv`, plus a timestamped copy). Nothing is uploaded if a check blocked or a board failed to build. Remove the `assert` when you're ready.

```python
assert False, (
    "Have you read the checks above, and is this leaderboard ready to publish?"
)

release.upload(STAGE, built, result)
```

## 5. Then

1. Set `CURRENT_PHASE = 2` in `hf_space/config.py` so the Space shows the interim tab.
2. Check the Space locally (`cd hf_space && python app.py`), then on the testing Space (push to `hf_testing`).
3. Merge to `main` to update the public Space.
